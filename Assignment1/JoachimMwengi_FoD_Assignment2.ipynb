### 2.1.4 Express E(T) as an infinite sum.

Given  we are dealing with discrete random variables, we know that For a discrete random variable $X$ with probability mass function $\mathbb{P}(X = x)$, the expectation is defined as: $$\mathbb{E}[X] = \sum_{x} x \, \mathbb{P}(X = x) $$

In our case, the formula takes the form:
$$\mathbb{E}[T] = \sum_{i=1}^{\infty} i \, \mathbb{P}(T = i)$$

Since: $$\mathbb{P}(T = i) = p \cdot (1-p)^{i-1}$$


Therefore by substitution:
$$\mathbb{E}[T] = \sum_{i=1}^{\infty} i \, p \cdot (1-p)^{i-1}$$

$$\mathbb{E}[T] = p \sum_{i=1}^{\infty} i \cdot (1-p)^{i-1}$$

### 2.1.5. Use the geometric series to derive a formula for $\sum_{i=1}^{\infty} i x^{i-1}$

Starting from the geometric series (for $|x| < 1$):
$$
\sum_{i=0}^{\infty} x^i = \frac{1}{1-x}
$$

Differentiate both sides with respect to $x$:

Left-hand side:
$$
\frac{d}{dx} \sum_{i=0}^{\infty} x^i = \sum_{i=1}^{\infty} i x^{i-1}
$$

Right-hand side:
$$
\frac{d}{dx}\left(\frac{1}{1-x}\right)
= \frac{d}{dx}\left((1-x)^{-1}\right)
$$

Using the chain rule:
$$
= -1 \cdot (1-x)^{-2} \cdot (-1)
= \frac{1}{(1-x)^2}
$$

Equating both expressions gives:
$$
\sum_{i=1}^{\infty} i x^{i-1} = \frac{1}{(1-x)^2}, \quad \text{for } |x| < 1.
$$

---

### 2.1.6. Compute the Expected Exit Time E[T]

Given our prior evaluation $$\mathbb{E}[T] = p \sum_{i=1}^{\infty} i \cdot (1-p)^{i-1}$$

And knowing the formula derived from the Geometric Series, $$\sum_{i=1}^{\infty} i x^{i-1} = \frac{1}{(1-x)^2}, \quad \text{for } |x| < 1.$$

Let $x = (1-p)$,

Then substitute,
$$\mathbb{E}[T] = p \cdot \sum_{i=1}^{\infty} i (1-p)^{i-1} = \frac{1}{(1-(1-p))^2}$$

Simplify: $$\mathbb{E}[T] = p \cdot \frac{1}{1-(1-p)^2} = p \cdot \frac{1}{(1 - 1 + p)^2} = p \cdot \frac{1}{p^2} $$

So, $$\mathbb{E}[T] = p \cdot \frac{1}{p^2}$$

$$ \therefore \mathbb{E}[T] = \frac{1}{p} $$

## Exercise 2.2. Fitting a spherical Gaussian to data

Consider a random variable |X⟩ ← Nd(|0⟩, 1) with spherical Gaussian distribution, mean |0⟩ and variance 1. Use d = 2 and d = 17.

#### 2.2.1 Generate a 1 000 point sample from |X⟩.
Hint: Use Python.

```python
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng(42)

def spherical_gaussian(d, n=1000):
    # Draw samples from a standard Normal distribution (mean=0, std-dev=1)
    return rng.standard_normal((n, d))

# Generating Samples for d = 2
gaussian_2D = spherical_gaussian(2, 1000)

# Generating Samples for d = 17
gaussian_17D = spherical_gaussian(17, 1000)

print(f"2D Gaussian Shape: {gaussian_2D.shape}")
print(f"17D Gaussian Shape: {gaussian_17D.shape}")
```

```python
# Visualize 2D Gaussian Sample
plt.scatter(gaussian_2D[:,0], gaussian_2D[:,1], s=5)
plt.title("Gaussian samples in d=2")
plt.axis("equal")
plt.show()
```

### 2.2.2. Compute the sample mean $^|μ1000⟩$ and the sample variance $σ^{2}_{1000}$.

```python
# Evaluate Sample Mean & Sample Variance

# 2D Gaussian Sample
mean_2D_gaussian = np.mean(gaussian_2D, axis=0)
var_2D_gaussian = np.var(gaussian_2D, axis=0)

# 17D Gaussian Sample
mean_17D_gaussian = np.mean(gaussian_17D, axis=0)
var_17D_gaussian = np.var(gaussian_17D, axis=0)

# 2D Gaussian 
print(f"d=2 Sample Mean = {mean_2D_gaussian}")
print(f"d=2 Sample Variance = {np.mean(var_2D_gaussian):.4f}\n")

# 17D Gaussian 
print(f"d=17 Sample Mean = {mean_17D_gaussian}")
print(f"d=17 Sample Variance = {np.mean(var_17D_gaussian):.4f}")
```

#### (a) What is the probability density for this sample?

### (a) Probability density for this sample

Let $$X_1, \dots, X_{1000} \in \mathbb{R}^d$$ be i.i.d. samples from a spherical Gaussian:
$$
X_i \sim \mathcal{N}_d(0, I)
$$

Since the samples are independent, the joint density factorises:

$$
p(X_1, \dots, X_{1000}) = \prod_{i=1}^{1000} p(X_i)
$$

Each term is:
$$
p(X_i) = \frac{1}{(2\pi)^{d/2}} \exp\left(-\frac{1}{2} X_i^T X_i \right)
$$

So the full joint density is:
$$
p(X_1,\dots,X_{1000})
=
\prod_{i=1}^{1000}
\frac{1}{(2\pi)^{d/2}}
\exp\left(-\frac{1}{2} X_i^T X_i \right)
$$

or further simplified into:
$$
p(X_1,\dots,X_{1000})
=
(2\pi)^{-1000d/2}
\exp\left(-\frac{1}{2}\sum_{i=1}^{1000} X_i^T X_i \right)
$$

#### (b) What is the probability density for this sample mean?

Define the sample mean:
$$
\hat{\mu}_{1000} = \frac{1}{1000} \sum_{i=1}^{1000} X_i
$$

Since a sum of Gaussian random variables is Gaussian:

$$
\hat{\mu}_{1000} \sim \mathcal{N}_d\left(0, \frac{1}{1000} I \right)
$$

Therefore its probability density is:
$$
p(\hat{\mu}) =
\frac{1}{(2\pi)^{d/2} \left|\frac{1}{1000}I\right|^{1/2}}
\exp\left(-\frac{1}{2} \hat{\mu}^T (1000 I)\hat{\mu}\right)
$$

Simplifying:
$$
p(\hat{\mu}) =
\frac{1000^{d/2}}{(2\pi)^{d/2}}
\exp\left(-\frac{1000}{2} \hat{\mu}^T \hat{\mu}\right)
$$

#### (c) What is the probability density for this sample variance?
