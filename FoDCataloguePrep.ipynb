### 1 Question
Find the SVD of the following matrix:

$$
A =
\begin{pmatrix}
3 & 2 & 2 \\
2 & 3 & −2
\end{pmatrix}
$$

### Solution

SVD of A is $ A = U \sum V^T$

The singular values are the square-roots of the eigenvalues of $AA^T$ & $ A^TA$. Since AA^T results in a 2x2 matrix, it is therefore simpler to compute:

#### 1. Compute $AA^T$

$$ AA^T = \begin{pmatrix}
3 & 2 & 2 \\
2 & 3 & −2
\end{pmatrix} \cdot \begin{pmatrix}
3 & 2 \\
2 & 3 \\
2 & -2
\end{pmatrix} $$

$$
AA^T = \begin{pmatrix}
17 & 8 \\
8 & 17
\end{pmatrix}
$$


#### 2. Compute $AA^T$'s Eigenvalues

$$  = det(AA^T - \lambda I) = 0 $$

$$  = \begin{vmatrix}
17-\lambda & 8 \\
8 & 17-\lambda
\end{vmatrix}  = 0 $$

$$  = (17 - \lambda)(17 - \lambda) - 8^2 = 0 $$
$$  = 17(17 - \lambda) - \lambda(17 - \lambda) - 64 = 0 $$
$$  = \lambda^2 - 34 \lambda + 225 = 0 $$
$$  = \lambda^2 - 9 \lambda - 25 \lambda + 225 = 0 $$
$$  = \lambda(\lambda - 9) - 25(\lambda - 9) = 0 $$
$$  = (\lambda - 25) (\lambda - 9) = 0 $$
$$  = \lambda _1 = 25$$
$$  = \lambda _2 = 9$$

#### 3. Compute Singular Values $ \sigma _i $

$$ \sigma _i = \sqrt{\lambda _i} $$

$$ \sigma _1 = \sqrt{25} = 5 $$
$$ \sigma _2 = \sqrt{9} = 3 $$

$$ \sum = \begin{pmatrix} 5 & 0 & 0 \\ 0 & 3 & 0 \end{pmatrix} $$


#### 4. Compute Eigenvectors of AA^T

$$ (AA^T - \lambda I)u_i $$

For $\lambda = 25$

$$ (\begin{vmatrix}
17 & 8 \\
8 & 17
\end{vmatrix}  - 25I)u_1 = 0 $$

$$ \begin{vmatrix}
-8 & 8 \\
8 & -8
\end{vmatrix}\begin{bmatrix} x \\ y \end{bmatrix} = 0 $$

$$ -8x + 8y = 0 $$
$$ x = y $$
$$ \begin{bmatrix} 1 \\ 1 \end{bmatrix} $$

Normalize $\sqrt{(1)^2 + (1)^2} = \sqrt{2}$

$$ u_1 = \frac{1}{\sqrt{2}} \begin{bmatrix} 1 \\ 1 \end{bmatrix}$$

For $\lambda = 9$

$$ (\begin{vmatrix}
8 & 8 \\
8 & 8
\end{vmatrix}  - 9I)u_2 = 0 $$

$$ \begin{vmatrix}
8 & 8 \\
8 & 8
\end{vmatrix}\begin{bmatrix} x \\ y \end{bmatrix} = 0 $$

$$ 8x + 8y = 0 $$
$$ x = -y $$
$$ \begin{bmatrix} 1 \\ -1 \end{bmatrix} $$

Normalize $\sqrt{(1)^2 + (-1)^2} = \sqrt{2}$

$$ u_2 = \frac{1}{\sqrt{2}} \begin{bmatrix} 1 \\ -1 \end{bmatrix}$$

$$ U = \begin{bmatrix} \frac{1}{\sqrt{2}} & \frac{1}{\sqrt{2}} \\ \frac{1}{\sqrt{2}} & \frac{-1}{\sqrt{2}} \end{bmatrix}$$


# Question 2

Show that for any $$A \in \mathbb{R}^{m \times n}$$, the matrices $$A^T A$$ and $$A A^T$$ possess the same **nonzero eigenvalues**.

#### 🧠 Strategy (What to focus on)

- Use the **definition of eigenvalues**
- Construct a mapping between eigenvectors
- Be explicit about the condition $$\lambda \neq 0$$
- Show both directions (completeness = higher marks)

---

#### ✏️ Proof

Assume:
$$
A^T A v = \lambda v, \quad v \neq 0
$$

with $$\lambda \neq 0$$.

---

##### Step 1: Multiply by $$A$$

$$
A (A^T A v) = A(\lambda v)
$$

$$
(A A^T)(A v) = \lambda (A v)
$$

---

##### Step 2: Define new vector

Let:
$$
u = A v
$$

Then:
$$
A A^T u = \lambda u
$$

---

##### Step 3: Justify nonzero condition

If $$\lambda \neq 0$$, then:
- $$v \neq 0$$
- ⇒ $$A v \neq 0$$ (otherwise LHS = 0 but RHS ≠ 0)

Thus $$u \neq 0$$.

---

##### Step 4: Reverse direction

Similarly, if:
$$
A A^T u = \lambda u
$$

Multiply by $$A^T$$:

$$
A^T (A A^T u) = \lambda A^T u
$$

$$
(A^T A)(A^T u) = \lambda (A^T u)
$$

Let:
$$
v = A^T u
$$

Then:
$$
A^T A v = \lambda v
$$

---

## Conclusion

The matrices $$A^T A$$ and $$A A^T$$ share the same **nonzero eigenvalues**.

---

## EXAM FOCUS

- MUST explicitly state:
  - $\lambda \neq 0 $ (critical!)
- Show:
  - Forward direction
  - Reverse direction
- Justify why vectors are **nonzero**
- Avoid skipping the step:
  > “If $$A v = 0$$ then contradiction”

👉 This is where most students lose marks.

---

## 🎯 High-Level Insight (Bonus Marks)

- $$A^T A$$ and $$A A^T$$ are both **positive semidefinite**
- Their nonzero eigenvalues are **squares of singular values of A**

### Question 3

#### Given

$$
x =
\begin{bmatrix}
1 \\ 2 \\ 3
\end{bmatrix}, \quad
y =
\begin{bmatrix}
-1 \\ -1 \\ 0
\end{bmatrix}
$$

Distance is defined via inner product:

$$
d(x,y) = \|x - y\| = \sqrt{\langle x-y, x-y \rangle}
$$

---

### (a) Standard Inner Product

#### Step 1: Compute difference

$$
x - y =
\begin{bmatrix}
2 \\ 3 \\ 3
\end{bmatrix}
$$

---

#### ✏️ Step 2: Compute norm

$$
\langle x-y, x-y \rangle = (x-y)^T (x-y)
$$

$$
= 2^2 + 3^2 + 3^2 = 4 + 9 + 9 = 22
$$

---

#### Final Answer

$$
d(x,y) = \sqrt{22}
$$

---

#### EXAM FOCUS

- Always:
  - Show vector subtraction explicitly
  - Expand dot product clearly
- Do NOT skip intermediate step

---

### (b) Weighted Inner Product

Inner product:

$$
\langle x, y \rangle = x^T A y
$$

where:

$$
A =
\begin{bmatrix}
2 & 1 & 0 \\
1 & 3 & -1 \\
0 & -1 & 2
\end{bmatrix}
$$

---

#### Step 1: Compute difference

$$
z = x - y =
\begin{bmatrix}
2 \\ 3 \\ 3
\end{bmatrix}
$$

---

#### Step 2: Compute $$A z$$

$$
A z =
\begin{bmatrix}
2 & 1 & 0 \\
1 & 3 & -1 \\
0 & -1 & 2
\end{bmatrix}
\begin{bmatrix}
2 \\ 3 \\ 3
\end{bmatrix}
=
\begin{bmatrix}
7 \\ 8 \\ 3
\end{bmatrix}
$$

---

#### Step 3: Compute quadratic form

$$
z^T A z =
\begin{bmatrix}
2 & 3 & 3
\end{bmatrix}
\begin{bmatrix}
7 \\ 8 \\ 3
\end{bmatrix}
$$

$$
= 2(7) + 3(8) + 3(3) = 14 + 24 + 9 = 47
$$

---

#### Final Answer

$$
d(x,y) = \sqrt{47}
$$

---

#### EXAM FOCUS

- Write distance formula correctly:
  $$
  d(x,y) = \sqrt{(x-y)^T A (x-y)}
  $$
- Show:
  - Matrix-vector multiplication
  - Final scalar computation
- Structure matters more than speed

---

#### High-Level Insight (Bonus Marks)

- This defines a **Mahalanobis-type distance**
- Matrix $$A$$ must be:
  - symmetric
  - positive definite

## Question 4

Consider the following bivariate distribution $p(x, y)$ of two discrete random variables X and Y .

| Y \\ X | x1 | x2 | x3 | x4 | x5 | p(y) |
|------|----|----|----|----|----|----|
| y1 | 0.01 | 0.02 | 0.03 | 0.10 | 0.10 | **0.26** |
| y2 | 0.05 | 0.10 | 0.05 | 0.07 | 0.20 | **0.47** |
| y3 | 0.10 | 0.05 | 0.03 | 0.05 | 0.04 | **0.27** |
| **p(x)** | **0.16** | **0.17** | **0.11** | **0.22** | **0.34** | **1** |


Compute:
    
    
    (a) The marginal distributions p(x) and p(y).


    (b) The conditional distributions p(x | Y = y1) and p(y | X = x3).

## Prerequisites

To solve this problem, you need to understand:

### 1. Joint Distribution
The table gives:
$$
p(x, y)
$$
which represents the probability of both \(X\) and \(Y\) occurring together.

---

### 2. Marginal Distributions

To eliminate one variable, sum over it:

- For \(p(x)\): sum down each column  
- For \(p(y)\): sum across each row  

$$
p(x) = \sum_y p(x,y), \quad p(y) = \sum_x p(x,y)
$$

---

### 3. Conditional Probability

$$
p(x|y) = \frac{p(x,y)}{p(y)}, \quad
p(y|x) = \frac{p(x,y)}{p(x)}
$$

---

## 📊 Given Table

| Y \\ X | x1 | x2 | x3 | x4 | x5 |
|------|----|----|----|----|----|
| y1 | 0.01 | 0.02 | 0.03 | 0.10 | 0.10 |
| y2 | 0.05 | 0.10 | 0.05 | 0.07 | 0.20 |
| y3 | 0.10 | 0.05 | 0.03 | 0.05 | 0.04 |

---

# (a) Marginal Distributions

## Compute \(p(x)\) (column sums)

$$
p(x_1)=0.01+0.05+0.10=0.16
$$

$$
p(x_2)=0.02+0.10+0.05=0.17
$$

$$
p(x_3)=0.03+0.05+0.03=0.11
$$

$$
p(x_4)=0.10+0.07+0.05=0.22
$$

$$
p(x_5)=0.10+0.20+0.04=0.34
$$

---

## 🔹 Compute \(p(y)\) (row sums)

$$
p(y_1)=0.01+0.02+0.03+0.10+0.10=0.26
$$

$$
p(y_2)=0.05+0.10+0.05+0.07+0.20=0.47
$$

$$
p(y_3)=0.10+0.05+0.03+0.05+0.04=0.27
$$

---

## Ans

$$
p(x) = (0.16,\ 0.17,\ 0.11,\ 0.22,\ 0.34)
$$

$$
p(y) = (0.26,\ 0.47,\ 0.27)
$$

---

## Sanity Check

$$
\sum p(x) = 1, \quad \sum p(y) = 1
$$

---

---

p(y_1) = 0.26
$$

$$
p(x_1|y_1)=\frac{0.01}{0.26}\approx 0.0385

$$
p(x_2|y_1)=\frac{0.02}{0.26}\approx 0.0769

$$

$$
p(x_5|y_1)=\frac{0.10}{0.26}\approx 0.3846
$$

## 2. $p(y \mid X = x_3)$

Column sum:
$$
p(x_3) = 0.11
$$

$$
p(y_1|x_3)=\frac{0.03}{0.11}\approx 0.2727
$$

$$

# Key Intuition

### Marginals:
“Add to remove a variable”

### Conditionals:
“Fix one variable → normalize so total = 1”

---

# 🎯 Exam Shortcut

- \(p(x|y_1)\): divide row by row sum  
- \(p(y|x_3)\): divide column by column sum
